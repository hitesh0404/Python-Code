```python
def hex_to_dec(number):
    result = 0
    for i in range(len(number)-1,-1,-1):
        ch = number[i]
        if ch in 'ABCDEFabcdef':
            if ord(ch) in range(97, 103):
                num = ord(ch)-87
            elif ord(ch) in range(65, 71):
                num = ord(ch)-55
        elif ch in '0123456789':
            num = int(ch)
        else:
            print("invalid hexadecimal literal")
            return None
        position = len(number)-i-1
        if position !=0:
            result += num * (16**position)
        else:
            result += num 
    return result
```

```python
class Human:
    def __init__(self,age):
        self.age = age
```

```python
class Employee(Human):
    def __init__(self,id:int,name:str,salary:float,age:int)->object:
        super().__init__(age)
        self.id=id
        self.name = name
        self.salary = salary
    def __str__(self):
        return self.name +" With id is "+ str(self.id)
    def add(self,a,b):
        return a+b
```

```python
class Manager(Employee):
    def __init__(self, id, name, salary,age):
        super().__init__(id, name, salary,age)
    def add(self,a,b):
        return a+b+1
```

```python
rakesh = Employee(1,"Rakesh",32312.3,9)
print(rakesh.add(1,2))
```
