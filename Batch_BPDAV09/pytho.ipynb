```python
num = [5,2,1,9,3,0]   
print(num)                  #  0 1 5 9 3 2
for i in range(0,len(num)):             #  0 1 2 3 4 5             1 
    for j in range(i+1,len(num)):       #                          3
        print(num[i],num[j],num)
        if num[i] > num[j]:             #  1 
            num[i],num[j] = num[j],num[i]
print(num)
```
