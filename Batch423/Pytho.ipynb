```python
def table_inside_file(file_name,num):
    with open(file_name,'x+') as my_data:
        for i in range(1,11):
            my_data.write(f'{num} X {i} = {i*num}\n')
        my_data.seek(25,0)
        print("my data is here\n",my_data.read())
        print('is redable',my_data.readable())
        print('is writable',my_data.writable())
print("hello")
```

```python
def write_table(file_name:str,num:int):
    if num==0:
        raise ArithmeticError("The table of zero is not allowed")
    with open(file_name+'.txt','x+') as f:
        for i in range(1,11):
            f.write(f'{i*num}\n')
        f.seek(0,0)
        print(f.readlines())
```

```python
class College:
    University = "Mumbai"
    def __init__(self,name,location):
        self.name = name
        self.location = location
    @classmethod
    def set_university(cls,name):
        cls.University= name
    @staticmethod
    def info():
        return "this is college class which hold one class level attribute university"
```

```python
viva = College("viva","virar")
bhavans = College("bhavans","Andheri")
print(viva.University,bhavans.University,College.University)
```

```python
viva.University = "Dubai"
print(viva.__dict__,viva.University,bhavans.University,College.University)
del viva.University
print(viva.__dict__,viva.University,bhavans.University,College.University)
viva.set_university("Dubai")
print(viva.__dict__,viva.University,bhavans.University,College.University)
print(viva.info(),'\n',College.info())
```
