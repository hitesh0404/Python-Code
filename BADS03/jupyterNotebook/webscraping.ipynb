```python
import requests
from bs4 import BeautifulSoup
re = requests.get("https://www.careers24.com/jobs/lc-mumbai/kw-data-developer/rmt-incl/")
```

```python
re.text
```

```python
bs = BeautifulSoup(re.text)
```

```python
jobs = []
for card in bs.select(".job-card"):
    info={} 
    info["title"]=card.select_one("a>h2").text
    info["location"] = card.select_one("ul>li:nth-of-type(1)").text.strip()
    info["job_type"] = card.select_one("ul>li:nth-of-type(2)").text.strip()[10:]
    date = card.select_one("ul>li:nth-of-type(3)").text.strip()[8:]
    info["posted_on"]  = date[0:date.find("\n")-1]
    info["days_left"] = card.select_one("ul>li>span").text
    jobs.append(info)
```

```python
import pandas as pd
pd.DataFrame(jobs)
```

```python
import pandas as pd

pd.DataFrame(jobs).to_excel("Data_Developer_job_in_mumbai.xlsx")
```
