# Numpy Array

- to perform mathmatical calculation on n dimentional data with efficiency and accuracy
- array is same as list which uses continue memory sequence for storing data of same data type
- n dimention(nd) array
- nd array is homogenious 
- nd array can be multidimention
- nd array can be of various DataType (e.g. int8 int16 float64)
- nd array support multiple methods
- nd array can be converted to different shape 

```python
import numpy as np
```

```python
d = [22,11,12,14,25]
print(d)
# [24,13,14,16,27]   
print(sum(d))
from functools import reduce
print(reduce(lambda a,b: a*b,d))
for index,data in enumerate(d) :
    d[index] = data+2
print(d)
```

```python
data1= [2,4,6,8,10,15]
print(np.std(data1),np.mean(data1),sep="\n")
data2 = [5,7,9,6,8,10]
print(np.std(data2),np.mean(data2),sep="\n")
```

```python
import numpy as np      # inbuild methods support
d = [22,11,12,14,25]
d = np.array(d)
print(d.sum())
print(d.prod())
print(d+2)
print(d.std())
print(d.mean())
print(d.var())
print(np.quantile(d,0.50))      # middle Value
```

## Broadcasting

```python
import numpy as np
d = np.array([1,2,3,4])  # applying on one dimensional array 
print(d)
print(d+2)
print(d*2)
print(d/2)
print(d//2)
print(d**2)
```

```python
d = np.array(range(1,5)).reshape(2,2) 
print(d)
print(d*2) # applying on two dimensional array 
```
