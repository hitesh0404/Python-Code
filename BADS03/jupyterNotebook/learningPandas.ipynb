```python
import pandas as pd
```

```python
pd.Series(["a","b","c"],index=[101,102,103])
```

```python
data = {
    "name": ["Hitesh","Manish","Amey","Nisha","Ashish"],
    "age" : [12,13,12,11,13]
}
df = pd.DataFrame(data,index=[101,102,103,104,105])
df["marks"] = [66,68,76,77,75]
df["age"] = df["age"]+10
df
```

```python
df.drop(["marks","age"],axis=1)
```

```python
df
```

```python
df.age         # can acces column if the column name doest not contain 
            #the special char or space
```

```python
df["name"]        # any column can be access with the name of the column
```

```python
df[["name","age"]]       # to access multiple columns
```

```python
df["age"]>=23     # when we try apply condition 
                    #it returns boolean array
```

```python
df[ df["age"]>=23 ]
```

```python
3 & 9       # 0111
            # 1001
         #and-------
            # 0001
```

```python
df[ (df["age"]>=23) | (df["marks"]>=70) ]
df[ (df["age"]>=23) & (df["marks"]>=70) ]
```

```python
df.query(" age >= 23 and marks >= 70")
df.query(" `age` >= 23 and `marks` >= 70 ")
```

```python
a = 23
m = 70
# we can use variable inside query with @
df.query(" `age` >= @a and `marks` >= @m ") 
```

```python
df.groupby("age").sum()
df.groupby("age").count()
```

```python
df.groupby("age").agg(["sum","count"])
```
