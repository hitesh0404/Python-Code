# print Function 
- print is use to print the output on console
- it has sep,end,file,flush perameter
- we can pass end and sep and other argument
### sep
- The sep parameter specifies the string to be inserted between multiple values passed to a single print() call. 
- Default value: A single space (' ').
- Purpose: To customize the separator between items. 
### end
- The end parameter specifies the string to be appended at the end of the output of the print() function. 
- Default value: A newline character ('\n').
- Purpose: To change what happens after the print statement finishes, such as keeping the output on the same line or adding custom text. 

```python
print(1,2,4,"hello",end="\n\n\n")
print(1,2,3,4,5,end=" this is the last")
```

### Arbitrary Arrguments
- more than one value can be consumed by one parameter
- that can be positional or keyword
- `*` is used with parameter to deal as a arbitrary 
- if its single then its postional or 
- if it is double then its keyword 

```python
def add(*num):
    print(num)
    total = 0
    for i in num:
        total = total + i
    return total
```

```python
add(2,3,4,5,3,4,2,1)
```

```python
a,*b,c = 10,1,21,2,2,1,1,1
b
```

```python
#. * for positional arbitrary
#. ** for keyword arbitrary

def greet(**data):
   
    name = data.get("name",None)
    age = data.get("age",None)

    if name and age:
        print(f"Hello my name is {name} and i am {age} years old")
    elif name:
        print(f"Hello my name is {name} ")
    else:
        print("Hello")

greet(name="Hitesh")
greet(name="Hitesh",age=12)
greet()
```

### map , filter
- takes function as an argument 
- apply a given function to each of the element 
- return generator object which can be converted to any other iterable like list,tuple,set
 

```python
def double(x):
    return x*2

data = [1,2,3,4,5,6]

result = []

for i in data:
    result.append(double(i))

print(result)

result = list(map(double,data))
print(result)
```

```python
def odd(x):
    return x%2 == 1

def even(x):
    return x%2 == 0

print(list(filter(odd,[1,2,3,4,5,6])))
print(list(filter(even,[1,2,3,4,5,6])))

```

## File Handling
- python can handle text file or binary file as well
- operation that can be performed in files are
- read
- write
- append
- create and write
- delete with os
- step to handle file
- 1. open or create then open
- 2. read/write/append
- 3. close

```python
from io import UnsupportedOperation
try:
    file = open("myfile.txt","r")
    print(file)
    print(file.read())       
    print(file.write("hey ")) #  error as of file mode is read
except UnsupportedOperation  as un:
    print("Error ",un)
except FileNotFoundError as fnf:
    print("Error " ,fnf)
finally:
    file.close()
```

- files mode
- 1. r  : Read (the content of file remain and the cursor point to start of the file)
- 2. w  : Write (the content of file vanish and the cursor point to start of the file)
- 3. w+ : Write and read (the content of file vanish and the cursor point to start of the file)
- 4. r+ : Read + Write (the content of file remain and the cursor point to start of the file)
- 5. a  : Append (the content of file remain and the cursor point to `end` of the file)
- 6. a+ : Append + Read (the content of file remain and the cursor point to `end` of the file)
- 7. x  : Create and write (if file exists then throws FileExistError)(the cursor point to start of the file)
- 8. x+ : Create then write + read  (if file exists then throws FileExistError)(the cursor point to start of the file)



```python
with open("myfile.txt","r") as file:
    print(file.read(2))
    file.seek(6,0)     # cookie: the step count    whence: 0 start     1 current     2 end
    print(file.read())
```
