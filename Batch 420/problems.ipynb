```python
def anagram(first,second):  #taste   sttte
    if len(first)!=len(second):
        return False
    letters = dict()
    for i in first:
        if i not in letters:
            letters[i] = 1
        else:
            letters[i] +=1
    for j in second:
        if letters.get(j,False):
            letters[j] -=1
        else:
            return False
    return True
```

```python
def append_string(s,index,value):
    new_index = 0
    for i in s:
        new_index += 1
        if i==' ':
            index-=1
        if index==0:
            break
    return s[0:new_index] + value + " " + s[new_index:]
```

```python
def new_append(s:str,index:int,value:str)->str:
    list_of_words = s.split()
    list_of_words.insert(index,value)
    result=''
    for i in list_of_words:
        result +=i + " "
    return result[:-1]
```

```python
def anti_clock(a:list):
    check = len(a[0])
    for i in a:
        if len(i)!=check:
            return False
    b=[]
    for j in range(len(a[0])-1,-1,-1):   #  2
        b.append(list())
        for i in range(len(a)):
            print(a[i][j],end="    ")
            b[len(a)-j-1].append(a[i][j])  # 3 - 2 -1  = 0         3 -1 -1   =1
        print()
    return b
```

```python
a= 'abcdef'
b= 'lmn'

def combite_at_alter_position(a:str,b:str)->str:
    length = len(a) if len(a)>len(b) else len(b)
    result = ''
    for i in range(length):
        if len(a)>i and len(b)>i:
            result += a[i]+b[i] 
        elif len(a)>i:
            result += a[i:]
            return result
        elif len(b)>i:
            result += b[i:]
            return result
        else:
            return result
    return result
print(combite_at_alter_position(a,b))
```

```python
def lets_start_count(num):
    def increase(increment_number=0):
        nonlocal num
        num += increment_number
        return num
    return increase
```

```python
def hash_problem(arr):
    fun = lambda x,n : int(x%n)
    def insert_into(value):
        nonlocal arr
        if value<0:
            return arr
        hash_code = fun(value,len(arr))
        while(True):
            if len(arr)<=hash_code:
                return "end"
            elif(arr[hash_code]):
                hash_code+=1
            else:
                arr[hash_code] = value
                return f'index ={hash_code} ,value = {value}'
    return insert_into
```
