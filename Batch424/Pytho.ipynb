### String Build in Methods
String has several build-in methods that can be used to manipulate strings. Here are some of the most
common ones:
python
# String Build in Methods
# 1. len() - Returns the length of the string
print(len("Hello World"))
# 2. lower() - Returns the string in lowercase
print("Hello World".lower())
# 3. upper() - Returns the string in uppercase
print("Hello World".upper())
# 4. title() - Returns the string with the first letter of each word capitalized
print("hello world".title())
# 5. swapcase() - Swaps the case of the string
print("Hello World".swapcase())
# 6. count() - Returns the number of occurrences of a substring
print("Hello World".count("l"))
# 7. find() - Returns the index of the first occurrence of a substring
print("Hello World".find("l"))
# 8. replace() - Replaces a substring with another substring
print("Hello World".replace("l", "x"))
\end{python}
The above code demonstrates the use of several string build-in methods in Python. These methods can be used
to manipulate strings in various ways.

```python
def rotate(arr:list,d,rotate_by:str = "right"):
    n = len(arr)
    if rotate_by.lower()=="right":
        for i in range(d):
            last = arr.pop()
            arr.insert(0,last)
        return arr
    elif rotate_by.lower()=="left":
        for i in range(d):
            first = arr.pop(0)
            arr.append(first)
        return arr
    else:
        return "Invalid rotate_by parameter. Please use 'right' or 'left'."
print(rotate([1,2,3],1))
```
