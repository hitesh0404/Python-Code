```python
import pandas as pd
from sqlalchemy import create_engine
```

```python
connection = create_engine(f'mysql://root:root@localhost:3308/classicmodels')
```

```python

customers_df = pd.read_sql_query("select * from customers",connection)
employees_df = pd.read_sql_query("select * from employees",connection)
offices_df = pd.read_sql_query("select * from offices",connection)
orderdetails_df = pd.read_sql_query("select * from orderdetails",connection)
orders_df = pd.read_sql_query("select * from orders",connection)
payments_df = pd.read_sql_query("select * from payments",connection)
productlines_df = pd.read_sql_query("select * from productlines",connection)
products_df = pd.read_sql_query("select * from products",connection)
customers_df.to_csv('customers.csv',index=False)
employees_df.to_csv('employees.csv',index=False)
offices_df.to_csv('offices.csv',index=False)
orderdetails_df.to_csv('orderdetails.csv',index=False)
orders_df.to_csv('orders.csv',index=False)
payments_df.to_csv('payments.csv',index=False)
productlines_df.to_csv('productlines.csv',index=False)
products_df.to_csv('products.csv',index=False)
```
