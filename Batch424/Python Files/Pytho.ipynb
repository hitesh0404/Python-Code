```python
def rotate(arr:list,d,rotate_by:str = "right"):
    n = len(arr)
    if rotate_by.lower()=="right":
        for i in range(d):
            last = arr.pop()
            arr.insert(0,last)
        return arr
    elif rotate_by.lower()=="left":
        for i in range(d):
            first = arr.pop(0)
            arr.append(first)
        return arr
    else:
        return "Invalid rotate_by parameter. Please use 'right' or 'left'."
print(rotate([1,2,3],1))
```

```python
def check_prime(num:int)->bool:  # num 8
    counter = 0
    for i in range(1,num+1):             #  1 2 3 4 5 6 7 8
          #1 self
        if num % i ==0:                  # 1  1   1         1
            counter+=1
    if counter <=2:
        return True
    else:
        return False
```

```python
def get_prime_with_range(start:int,end:int)->list:
    result = []
    for num in range(start,end):
        counter = 0
        for i in range(1,num+1):             #  1 2 3 4 5 6 7 8
            #1 self
            if num % i ==0:                  # 1  1   1         1
                counter+=1
        if counter <=2:
            result.append(num)
    return result
```
